```python
%pip install -q scikit-learn pandas pypdf gradio
import gradio as gr
import pandas as pd
from pypdf import PdfReader
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# نصوص افتراضية جاهزة للتجربة حتى لو ما رفعت ملفات
DEFAULT_DOCS = [
    {
        "source": "nlp_intro.txt",
        "location": "ملف تجريبي",
        "text": (
            "معالجة اللغات الطبيعية تهتم بفهم وتحليل النصوص البشرية. تُستخدم"
            " تقنيات مثل TF-IDF وحساب تشابه الجيب التمام في بناء محركات البحث."
        ),
    },
    {
        "source": "search_engines.txt",
        "location": "ملف تجريبي",
        "text": (
            "محركات البحث تعتمد على فهرسة الوثائق وترتيب النتائج حسب الصلة."
            " تحليل النصوص يساعد في العثور على الكلمات المفتاحية بسرعة."
        ),
    },
    {
        "source": "arabic_nlp.txt",
        "location": "ملف تجريبي",
        "text": (
            "تواجه معالجة اللغة العربية تحديات مثل تعدد الأشكال والتشكيل، ولكن"
            " تطبيق خوارزميات التصنيف والبحث يعطي نتائج عملية مفيدة."
        ),
    },
]


def extract_documents(files):
  """استخراج النصوص من الملفات المرفوعة (TXT أو PDF)"""
  if not files:
    return DEFAULT_DOCS

  documents = []
  for file_obj in files:
    file_path = file_obj.name
    filename = file_path.split("/")[-1].split("\\")[-1]

    if filename.lower().endswith(".txt"):
      with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        text = f.read().strip()
      if text:
        documents.append({"source": filename, "location": "ملف نصي", "text": text})

    elif filename.lower().endswith(".pdf"):
      reader = PdfReader(file_path)
      for page_num, page in enumerate(reader.pages, start=1):
        page_text = (page.extract_text() or "").strip()
        if page_text:
          documents.append({
              "source": filename,
              "location": f"صفحة {page_num}",
              "text": page_text,
          })

  return documents if documents else DEFAULT_DOCS


def search_interface(files, query, top_k):
  if not query or not query.strip():
    return pd.DataFrame(
        {"تنبيه": ["الرجاء كتابة كلمة أو عبارة للبحث عنها."]}
    )

  docs = extract_documents(files)
  corpus = [d["text"] for d in docs]

  vectorizer = TfidfVectorizer(lowercase=False, token_pattern=r"(?u)\b\w+\b")
  tfidf_matrix = vectorizer.fit_transform(corpus)
  query_vec = vectorizer.transform([query])

  similarities = cosine_similarity(query_vec, tfidf_matrix).flatten()

  results = []
  for idx, score in enumerate(similarities):
    doc = docs[idx]
    found_exact = query.lower() in doc["text"].lower()
    if score > 0 or found_exact:
      snippet = doc["text"].replace("\n", " ")
      if len(snippet) > 160:
        snippet = snippet[:160] + "..."
      results.append({
          "اسم الملف": doc["source"],
          "الموقع": doc["location"],
          "درجة التشابه (Cosine)": round(float(score), 4),
          "مقتطف من النص": snippet,
      })

  if not results:
    return pd.DataFrame({"النتيجة": ["لم يتم العثور على وثائق مطابقة للبحث."]})

  df = pd.DataFrame(results)
  df = df.sort_values(by="درجة التشابه (Cosine)", ascending=False).head(
      int(top_k)
  )
  return df.reset_index(drop=True)


# بناء واجهة البرنامج
with gr.Blocks(title="محرك بحث الوثائق - NLP") as demo:
  gr.Markdown("# 🔍 محرك بحث بسيط للوثائق (TF-IDF & Cosine Similarity)")
  gr.Markdown(
      "ارفع ملفات **PDF** أو **TXT** (أو اتركها فارغة لاستخدام الملفات"
      " التجريبية)، ثم اكتب كلمة البحث."
  )

  with gr.Row():
    file_input = gr.File(
        label="ارفع ملفاتك هنا (PDF / TXT)",
        file_count="multiple",
        file_types=[".txt", ".pdf"],
    )
    with gr.Column():
      query_input = gr.Textbox(
          label="كلمة أو عبارة البحث", placeholder="مثال: معالجة اللغات أو TF-IDF"
      )
      top_k_slider = gr.Slider(
          minimum=1, maximum=10, value=5, step=1, label="عدد النتائج المعروضة"
      )
      search_btn = gr.Button("ابحث الآن", variant="primary")

  output_table = gr.Dataframe(label="نتائج البحث", interactive=False)

  search_btn.click(
      fn=search_interface,
      inputs=[file_input, query_input, top_k_slider],
      outputs=output_table,
  )

demo.launch(share=True)
```
